### List vs numpy

```python
import numpy as np
```

```python
py_list = [1,2,3]
print(f"Python list multiplication {py_list*2}")
```

```python
np_array = np.array(py_list)
print(f"Python list multiplication {np_array*2}")
```

```python
import time

start = time.time()
py_list = [i*2 for i in range(1000000)]
print(f"List operation time: {time.time()-start}")
```

```python
%%time
py_list = [i*2 for i in range(1000000)]
print(f"List operation time: ")
```

```python
%%time
np_array = np.arange(1000000)*2
print(f"Numpy operation time: ")
```

### Creating array from scratch

```python
zeros = np.zeros((3,4)) 
print("Zeros array: \n",zeros)


ones = np.ones((2,3)) 
print("Ones array: \n",ones)

full = np.full((2,2),7) 
print("full array: \n",full)
```

```python
random = np.random.random([2,3])
print("random array: \n", random)
```

```python
sequence = np.arange(0,11,2)
print("Sequence array: \n", sequence)
```

### Vector, Matrix and Tensor

```python
vector = np.array([1,2,3])
print("Vector: ",vector)
```

```python
matrix = np.array([[1,2,3],[4,5,6]])
print("matrix: ",matrix)
```

```python
tensor = np.array([[[1,2],[3,4]],[[5,6],[7,8]]])
print("tensor: ",tensor)
```

### Array Properties

```python
arr = np.array([[1,2,3],
                [4,5,6]])

print("Shape ", arr.shape)
print("Dimension ", arr.ndim)
print("Size ", arr.size)
print("DType ", arr.dtype)
```

### Array Reshaping

```python
arr = np.arange(12)

print(f"Original Array: {arr}")

reshaped = arr.reshape((3,4))
print(f"Reshaped Array: {reshaped}")


flattened = reshaped.flatten()
print(f"Flattened Array: {flattened}")


raveled = reshaped.ravel()
print(f"Raveled  Array: {raveled}")

```
