Calling the Chat Completions API

```python
import os
import requests
import json
```

```python
from dotenv import load_dotenv,find_dotenv

_: bool = load_dotenv(find_dotenv())
```

```python
client = os.environ["OPEN_API_KEY"]
```

```python
url:str = "https://api.openai.com/v1/chat/completions"
```

```python
prompt:str = "Why is Hunzza a good tourist destination?"
```

```python
headers = {
    "Content-Type": "application/json",
    "Authorization": f'Bearer {client}'
}
```

```python
data = {
    "model": "gpt-3.5-turbo-1106",
    "messages":[{"role":"user","content":prompt}],
    "temperature": 0.7,
}
```

```python
response = requests.post(url, headers=headers, json=data, timeout=100)

if response.status_code != 200:
    print("Request Failed with status code:", response.status_code)

result = response.json()
```

```python
print(json.dumps(result['choices'][0]['message']['content'], indent=2))
```
