```python
from typing import List
import PyPDF2

def read_pdf(file_path: str) -> List[str]:
    with open(file_path, 'rb') as file:  # The 'b' in 'rb' stands for binary mode
        reader: PyPDF2.PdfReader = PyPDF2.PdfReader(file)
        text_content: List[str] = [page.extract_text() for page in reader.pages]
        return text_content

pages: list[str] = read_pdf("./mypdf.pdf")
pages
```

### Modes For Opening Files
- r - Read a file ( no warning ) - this is the default
- w - Write to a file (previous contents removed)
- a - Append to a file (previous contents not removed)
- r+ - Read and write4 to a file (writing based on cursor) - r,r+ work with existing file


```python
with open("haiku.txt","r") as file:
    print(file.read())
```

```python
import os

def list_dir(fld: str) -> None:
    abs_path = os.path.abspath(fld)
    print(f"🔍 Looking in: {abs_path}")

    if not os.path.exists(abs_path):
        print(f"❌ Error: Directory '{abs_path}' does not exist.")
        return
    if not os.path.isdir(abs_path):
        print(f"❌ Error: '{abs_path}' is not a directory.")
        return

    for fn in os.listdir(abs_path):
        print(fn)

# Use current directory
list_dir('../12_Error_and_File_Handling')
```

```python
import os

def ends_with(fld:str,search:str)->None:
    for fn in os.listdir(fld):
        if fn.endswith(search):
            print(fn)

def starts_with(fld:str,search:str)->None:
    for fn in os.listdir(fld):
        if fn.startswith(search):
            print(fn)


# ends_with("../12_Error_and_File_Handling",'.txt')
starts_with("../12_Error_and_File_Handling",'abc')
```

```python
import os,fnmatch

def match(fld:str,search:str)->None:
    for fn in os.listdir(fld):
        if fnmatch.fnmatch(fn,search):
            print(fn)

match("../12_Error_and_File_Handling",'*a*')
```

```python
from pathlib import Path


def glob_match(fld:str,search:str)->None:
    p = Path(fld)
    for n in p.glob(search):
        print(n)

glob_match("../12_Error_and_File_Handling","*.txt")
```

```python
import os 
from datetime import datetime,timezone

def get_date(timestamp:float)->str:
     return datetime.fromtimestamp(timestamp, tz=timezone.utc).strftime('%d %b %Y')

def get_file_attrs(fld:str):
     with os.scandir(fld) as dir:
          for f in dir:
               if f.is_file():
                    inf = f.stat()
                    print(f"Modified {get_date(inf.st_mtime)} {f.name}")


get_file_attrs("../12_Error_and_File_Handling")
```

```python
import os 
def traverse(fld:str)->None:
    for fldpath, dirs, fls in os.walk(fld):
        print(f'Folder: {fldpath} || {dirs}')
        for fn in fls:
            print(f'\t{fn}')


traverse("../../python")
```
